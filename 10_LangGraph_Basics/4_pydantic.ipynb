### Pydantic Data Validation

```python
from langgraph.graph import StateGraph,START,END
from pydantic import BaseModel 
```

```python
class State(BaseModel):
    name:str
```

```python
 

# 2. Define your node function
def example_node(state: State) -> dict:
    # Perform state updates here
    return {"name","Hello"} 
```

```python
# 3. Build the StateGraph
builder = StateGraph(State)
# 4. Add nodes and edges
builder.add_node("example_node", example_node)
builder.add_edge(START, "example_node")
builder.add_edge("example_node", END)
# 5. Compile the graph
graph = builder.compile()
```

```python
# To visualize (in Jupyter / IPython):
from IPython.display import Image, display
display(Image(graph.get_graph().draw_mermaid_png()))

# To invoke the graph:
result = graph.invoke({"name":"Rishabh"})
```
